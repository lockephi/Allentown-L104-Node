# Deutsch-Jozsa Algorithm Analysis
## L104 Quantum Advantage Demonstration

This notebook demonstrates exponential quantum speedup using the Deutsch-Jozsa algorithm.

```python
import sys
sys.path.append('..')
from oracle import DeutschJozsaOracle
import matplotlib.pyplot as plt
import numpy as np

# Initialize quantum resources
print("L104 Quantum Processing Unit Initialized")
```

```python
def run_experiment(n_qubits_list):
    """Compare classical vs quantum query complexity"""
    classical_queries = []
    quantum_queries = []
    
    for n in n_qubits_list:
        # Classical worst-case: 2^(n-1) + 1 queries
        classical_queries.append(2**(n-1) + 1)
        # Quantum: always 1 query
        quantum_queries.append(1)
        
        # Demonstrate actual quantum execution
        oracle = DeutschJozsaOracle(n_qubits=n, oracle_type='balanced')
        result = oracle.execute(shots=100)
        print(f"n={n}: Quantum determined {result['determination']} in 1 query")
    
    return classical_queries, quantum_queries
```

```python
# Run comparative analysis
n_values = [1, 2, 3, 4, 5]
classical, quantum = run_experiment(n_values)

# Plot results
plt.figure(figsize=(10, 6))
plt.plot(n_values, classical, 'r-', label='Classical (worst-case)', linewidth=2)
plt.plot(n_values, quantum, 'b-', label='Quantum (Deutsch-Jozsa)', linewidth=2)
plt.xlabel('Number of Qubits (n)', fontsize=12)
plt.ylabel('Queries Required', fontsize=12)
plt.title('Exponential Quantum Advantage: Deutsch-Jozsa Algorithm', fontsize=14)
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.legend()
plt.savefig('quantum_advantage.png', dpi=300)
plt.show()
```
